```python
%sql
CREATE OR REPLACE VIEW
    workspace.lakehouse_demo.v_customers_by_location
AS
SELECT
    city,
    country,
    COUNT(*) AS customer_count
FROM workspace.lakehouse_demo.silver_customers
GROUP BY city, country;
```

```python
%sql
SELECT *
FROM workspace.lakehouse_demo.v_customers_by_location
ORDER BY customer_count DESC;
```
