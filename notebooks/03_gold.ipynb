```sql
%sql
CREATE OR REPLACE TABLE workspace.lakehouse_demo.gold_sales_daily
USING DELTA
AS
SELECT
    order_date,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS units_sold,
    SUM(order_amount * quantity) AS total_revenue
FROM workspace.lakehouse_demo.silver_orders_enriched
WHERE status = 'success'
GROUP BY order_date;
```

```sql
%sql
CREATE OR REPLACE TABLE workspace.lakehouse_demo.gold_sales_by_product
USING DELTA
AS
SELECT
    product_id,
    product_name,
    category,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS units_sold,
    SUM(order_amount * quantity) AS revenue
FROM workspace.lakehouse_demo.silver_orders_enriched
WHERE status = 'success'
GROUP BY product_id, product_name, category;
```

```sql
%sql
SELECT
    order_date,
    total_orders,
    units_sold,
    ROUND(total_revenue, 2) AS total_revenue
FROM workspace.lakehouse_demo.gold_sales_daily
ORDER BY order_date;
```

```sql
SELECT
    product_id,
    product_name,
    category,
    units_sold,
    ROUND(revenue, 2) AS revenue
FROM workspace.lakehouse_demo.gold_sales_by_product
ORDER BY revenue DESC
LIMIT 10;
```

```sql
SELECT 'silver_success' AS origine,
       SUM(order_amount * quantity) AS revenu
FROM workspace.lakehouse_demo.silver_orders_enriched
WHERE status = 'success'

UNION ALL

SELECT 'gold_daily', SUM(total_revenue)
FROM workspace.lakehouse_demo.gold_sales_daily

UNION ALL

SELECT 'gold_products', SUM(revenue)
FROM workspace.lakehouse_demo.gold_sales_by_product;
```
