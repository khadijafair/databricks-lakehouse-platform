```python
dbutils.widgets.dropdown(
    "table_name",
    "gold_sales_daily",
    ["gold_sales_daily", "gold_sales_by_product"],
    "Table à contrôler"
)

dbutils.widgets.text(
    "min_rows",
    "1",
    "Nombre minimum de lignes"
)
```

```python
table_name = dbutils.widgets.get("table_name")
min_rows = int(dbutils.widgets.get("min_rows"))

allowed_tables = {"gold_sales_daily", "gold_sales_by_product"}

if table_name not in allowed_tables:
    raise ValueError("Table non autorisée pour cet exercice.")

if min_rows < 0:
    raise ValueError("Le minimum doit être positif ou nul.")

full_name = f"workspace.lakehouse_demo.{table_name}"
row_count = spark.table(full_name).count()

print(f"Table : {full_name}")
print(f"Lignes trouvées : {row_count}")
print(f"Minimum attendu : {min_rows}")

if row_count < min_rows:
    raise ValueError(
        f"Contrôle échoué : {row_count} lignes, "
        f"minimum attendu {min_rows}."
    )

print("✅ Contrôle réussi")
```
